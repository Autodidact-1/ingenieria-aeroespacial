## **5 Funciones trascendentes**

### **5.1 Funciones circulares**

En muchas situaciones reales es necesario determinar posiciones y distancias sobre trayectorias curvas. Un ejemplo de ello ocurre en una pista de atletismo circular, donde la ubicación exacta de una meta no siempre coincide con el punto de partida ni puede determinarse mediante mediciones directas sobre la pista.

Consideremos una pista circular de $30$ metros de radio en la que la línea de salida permanece fija. Se desea organizar una carrera de $1000$ metros y determinar con precisión el punto de llegada. Para resolver este problema será necesario calcular la longitud de la circunferencia, establecer cuántas vueltas completas realiza el atleta y relacionar la longitud recorrida sobre el arco con las distancias medidas en línea recta. De esta manera, se aplicarán conceptos geométricos como circunferencia, arco, ángulo central y cuerda para localizar exactamente la posición de la meta.

**Resolución de la situacion del problema**:

La longitud de la circuferencia nos ayudará a entender la cantidad de vueltas que singnifica recorrer $1000$ metros.

$$l_c{}=2\pi{}r=2\pi{}\cdot{}(30m)\approx{}188.5m$$

Dividimos la distancia total por la longitud de una vuelta.

$$n=\frac{L}{l_{c}}\approx{}\frac{1000m}{188.5m}\approx{}5.30$$

El atleta habrá dado aproximadamente 5,31 vueltas, es decir $5$ vueltas completas más un $30\%$ de la sexta vuelta.

Ubicar aproximadamente el punto de llegada después de 5 vueltas completas, el corredor vuelve a la línea de salida, luego debe recorrer...

$$L-5(l_{c})\approx{}1000m-5(188.5m)\approx{}57.5m$$

Como una vuelta mide $188.5m$

$$\frac{57.5m}{188.5m}\cdot{}360°\approx{}110°$$

Por lo tanto, el punto de llegada se encuentra aproximadamente a **$110°$ desde la línea de salida**, siguiendo el sentido de la carrera.

Ubicar exactamente el punto de llegada $(x,y)$ ay de definir $x$ e $y$ con ayuda de la trigonometria y el radio de la pista.

$$x=r\cdot{}cos(\theta{})$$
$$y=r\cdot{}sen(\theta{})$$

Para ubicar el punto exacto de llegada hay que encontrar el valor angulo $\theta{}_{rad}$ en radianes.

$$\theta{}_{rad}=\frac{arco}{radio}=\frac{l_c(n-5)}{r}\approx{}\frac{188.5m(5.30-5)}{30m}\approx{}1.91$$
$$\theta{}_{rad}=2\pi{}(n-5)\approx{}2\pi{}(5.30-5)\approx{}1.91$$

Sustituimos el valor del angulo en las expreciones de $x$ e $y$

$$x=r\cdot{}cos(\theta{}_{rad})\approx{}30m\cdot{}cos(1.91)\approx{}-10.19$$
$$y=r\cdot{}sen(\theta{}_{rad})\approx{}30m\cdot{}sen(1.91)\approx{}28.21$$

El punto exacto de llegada concluye en $(-10.19,28.21)$ con respecto al centro $(0,0)$.

En la situación anterior hemos comenzado con un número real ($1000$) al cual le hemos asociado un punto sobre la circunferencia de radio $30$ y con un sistema de ejes conveniente encontramos las coordenadas del punto. Para hacerlo, hemos asociado al número con un recorrido sobre la circunferencia y recurrimos a la trigonometría para calcular las coordenadas.

En lo que sigue, daremos una vision de las funciones trigonometricas un tanto diferente a la que ustedes han conocido, en la cual las funciones se definian sobre angulos. Nuestro proposito será definir las funciones seno, coseno y tangente como funciones numéricas, y estudiar algunas de sus propiedades con los métodos desarrollados en las Secciones anteriores. Esta manera de ver a las funciones trigonométricas será de gran importancia en las materias que siguen, tanto de Matemática como de Fisica. Un ejemplo de esto es la descripción del movimiento circular, que estudiaremos un poco más adelante.

#### **5.1.1 La circuferencia unitaria**

El conjunto de puntos del plano situados a distancia $1$ del origen de la circuferencia de radio $1$, centrada en el origen. Hemos aprendido que la ecuación de esta circuferencia es $x^2+y^2=1$.

***
$\color{red}\text{Ejemplo}$:

El punto $P\left( \frac{\sqrt{3}}{3},\frac{\sqrt{2}}{\sqrt{3}} \right)$ está sobre la circunferencia unitaria, puesto que
$$\left( \frac{\sqrt{3}}{3} \right)^2+\left( \frac{\sqrt{2}}{\sqrt{3}} \right)^2=\frac{3}{9}+\frac{2}{3}=\frac{1+2}{3}=1$$
***

Los ejes coordenados determinan cuatro regiones del plano llamadas **cuadrantes**. 



Los cuadrantes se pueden describirse como:

1. Primer cuadrante: $C_1 = \{(x, y): x > 0 \text{ e } y > 0\}$.

2. Segundo cuadrante: $C_2 = \{(x, y): x < 0 \text{ e } y > 0\}$.

3. Tercer cuadrante: $C_3 = \{(x, y): x < 0 \text{ e } y < 0\}$.

4. Cuarto cuadrante: $C_4 = \{(x, y): x > 0 \text{ e } y < 0\}$.

**Puntos terminales en la circunferencia unitaria**.

Supongamos que $t$ es un número real positivo. Marquemos una distancia igual a $t$ a lo largo de la circunferencia unitaria de la siguiente manera: empezamos en el punto $(1, 0)$ y nos movemos $t$ unidades sobre la circunferencia en el sentido opuesto al de las agujas del reloj. De esta manera llegamos a un punto $P (x, y)$.

Para $t$ negativo, comenzando en el mismo punto recorremos $|t|$ unidades sobre la circunferencia, pero en sentido contrario al anterior.

De esta manera, para cualquier número real $t$ obtenemos un punto $P(x,y)$ sobre la circunferencia unitaria. El punto $P(x,y)$ así obtenido se llama punto terminal determinado por el número real $t$.

La longitud de la circunferencia unitaria es $C=2\pi{}\cdot{}1=2\pi{}$. Entonces si un punto empieza su recorrido en $(1,0)$ y se mueve una distancia igual a $2\pi{}$ en sentido antihorario, llegará nuevamente a $(1, 0)$. Si recorre una distancia igual a $\frac{1}{2}\cdot{}2\pi{}=\pi{}$ el punto terminal será en este caso el $(−,0)$. De la misma forma el punto terminal para $\pi{}/2$ (que es la cuarta parte de la longitud total de la circunferencia) será el $(0, 1)$.

**Puntos terminales de números reales particulares**.

Calculemos el punto terminal $P(x,y)$ para $t=\pi{}/4$. Dicho punto está a la misma distancia desde $(1, 0)$ que de $(0, 1)$ a lo largo de la circunferencia unitaria.

1. Puesto que la circunferencia unitaria es simétrica respecto a la recta $y=x$, deducimos que $P(x,y)$ está en la recta $y=x$. $P$ será el punto de intersección entre la circunferencia de ecuación $x^2+y^2=1$ y la recta de ecuación $y=x$, en el primer cuadrante $C_1$.

2. Determinamos las coordenadas $(x,y)$ resolviendo en el esquema siguiente el sistema con las ecuaciones de la recta y la circunferencia 
    $$\begin{cases} x^2+y^2=1 \\ y=x \end{cases}$$
    Como $y=x$ sustituimos a $x$ a la ecuación $x^2+y^2=1$ 
    $$x^2+x^2=1$$
    $$2x^2=1$$
    $$x^2 =\frac{1}{2}$$ $$|x|=\frac{1}{\sqrt{2}}$$
    Podemos observar que $x$ puede tomar un valor negativo y positivo $\pm{}\frac{1}{\sqrt{2}}$, como $t=\pi{}/4$ se encuentra en el primer cuadrante $x > 0$ e $y > 0$ ambos tomaran valores positivos, y como $y=x$ ambos toman el mimo valor.
    $$x=\frac{1}{\sqrt{2}};\quad{} y=\frac{1}{\sqrt{2}}$$

3. Por lo tanto el punto terminal determinado por $t=\pi{}/4$ es $P\left( \frac{1}{\sqrt{2}},\frac{1}{\sqrt{2}} \right)$.

Para obtener los puntos terminales determinados por $t=\pi{}/6$ y $t=\pi{}/3$ pueden utilizarse métodos similares.

1. **Para $t=\pi{}/6$ (30°)**:
   Si reflejamos el punto $P(x, y)$ respecto al eje $x$, obtenemos $P'(x, -y)$. La distancia entre $P$ y $P'$ es $2y$. Al ser el ángulo central $2 \cdot (\pi/6) = \pi/3$ (60°), el triángulo formado por el origen $O(0,0)$, $P$ y $P'$ es **equilátero**. Como los lados $OP$ y $OP'$ miden $1$ (radio), el lado $PP'$ también debe medir $1$.
   $$2y = 1 \implies y = \frac{1}{2}$$
   Sustituyendo en la ecuación de la circunferencia $x^2 + y^2 = 1$:
   $$x^2 + \left(\frac{1}{2}\right)^2 = 1 \implies x^2 = \frac{3}{4} \implies x = \frac{\sqrt{3}}{2}$$
   Por lo tanto, el punto terminal es $P\left( \frac{\sqrt{3}}{2}, \frac{1}{2} \right)$.

2. **Para $t=\pi{}/3$ (60°)**:
   Este caso es simétrico al anterior respecto a la recta $y=x$. Aquí, la distancia al eje $y$ es la que resulta ser $1/2$ (usando un razonamiento análogo de triángulo equilátero pero reflejando respecto al eje $y$).
   $$x = \frac{1}{2}$$
   Sustituyendo en $x^2 + y^2 = 1$:
   $$\left(\frac{1}{2}\right)^2 + y^2 = 1 \implies y^2 = \frac{3}{4} \implies y = \frac{\sqrt{3}}{2}$$
   Por lo tanto, el punto terminal es $P\left( \frac{1}{2}, \frac{\sqrt{3}}{2} \right)$.

| $t$ | Punto terminal determinado por $t$ |
|:-:|:-:|
| $0$ | $(1,0)$ |
| $\pi{}/6$ | $\left( \frac{\sqrt{3}}{2}, \frac{1}{2} \right)$ |
| $\pi{}/4$ | $\left( \frac{\sqrt{2}}{2}, \frac{\sqrt{2}}{2} \right)$ |
| $\pi{}/3$ | $\left( \frac{1}{2}, \frac{\sqrt{3}}{2} \right)$ |
| $\pi{}/2$ | $(0,1)$ |
