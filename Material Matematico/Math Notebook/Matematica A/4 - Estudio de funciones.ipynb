## **4 Estudio de Funciones**

### **4.1 Funciones derivables y no derivables**

Hemos interpretado geométricamente a la derivada de una función $f(x)$ en un valor a como la pendiente de la recta tangente a la gráfica de $f(x)$ por el punto $(a, f(a))$. Ahora bien ¿siempre existirá esa tangente? A la derivada la hemos definido como un límite (el límite del cociente incremental cuando el incremento de la variable independiente tiende a $0$), y la existencia o no de la derivada se corresponde con la existencia o no de ese limite.

***
$\color{blue}\text{Ejemplo}$:
Analizar en qué valores es derivable la función valor absoluto $f(x) = |x|$.

Recordemos que la función valor absoluto es una función definida a trozos y que $x = 0$ es el punto de pegado. Podemos ver su gráfica al margen.

Es claro que si $x < 0$ la gráfica tiene tangente por el punto $(x, −x)$. Y esa tangente es $y = −x$, recta que coincide con la gráfica de la función para $x < 0$. Lo mismo sucederá para $x > 0$. Pero algo extraño pasa en $x = 0$; la gráfica tiene allí un punto anguloso y no parece posible apoyar allí una tangente.

Veamos, el cociente incremental en $x = 0$ de la función valor absoluto es:

$$\frac{f(0 + h) −f(0)}{h} =\frac{|0 + h|−|0|}{h}=\frac{|h|}{h}$$

Y queremos calcular el limite de esa expresión cuando $h\to{}0$. Como, justamente, $x = 0$ es el punto de pegado; conviene entonces estudiar los limites laterales del cociente incremental:

$$\lim_{h\to{}0^+}\frac{f(0 + h) −f(0)}{h}=\lim_{h\to{}0^+}\frac{|h|}{h}=\lim_{\substack{h\to{}0 \\ h>0}}\frac{|h|}{h}=\lim_{h\to{}0}\frac{h}{h}=1$$

$$\lim_{h\to{}0^−}\frac{f(0 + h) −f(0)}{h}=\lim_{h\to{}0^−}\frac{|h|}{h}=\lim_{\substack{h\to{}0 \\ h<0}}\frac{|h|}{h}=\lim_{h\to{}0}\frac{−h}{h}=−1$$

Comprobamos que los limites laterales son distintos y, por lo tanto, no existe $\displaystyle{\lim_{h\to{}0}}\frac{f(0+h)−f(0)}{h}$. Podemos concluir que la función no es derivable en $x = 0$.
***

Hemos encontrado que hay funciones (aun funciones sencillas y usuales como las del ejemplo anterior) que no son derivables en algunos valores. Para terminar con el ejemplo del valor absoluto notemos que si bien no es derivable en $x = 0$ sí es continua en ese valor.

Haciendo referencia al Teorema de 
> *"Si una función es derivable en un punto entonces es continua en dicho punto."*

allí vimos que una función derivable es necesariamente continua. El ejemplo anterior nos muestra que la recíproca de esa afirmación no es cierta (el valor absoluto es continuo en $x = 0$ pero no es derivable en ese valor).

Ya podemos describir gráficamente dos situaciones en las que una función no es derivable en un valor a:

* La gráfica de $f(x)$ tiene un punto anguloso en $(a, f(a))$.

* La gráfica de $f(x)$ se corta en $x = a$.

Veamos un ejemplo de una situación diferente a las anteriores:

***
$\color{blue}\text{Ejemplo}$:

Ver que la función $f(x)=\sqrt[3]{x}$ no es derivable en $x = 0$.

En efecto, planteemos el limite del cociente incremental:

$$\lim_{h\to{}0}\frac{f(0+h)−f(0)}{h}=\lim_{h\to{}0}\frac{\sqrt[3]{0+h}-\sqrt[3]{0}}{h}=\lim_{h\to{}0}\frac{\sqrt[3]{h}}{h}=\lim_{h\to{}0}\frac{\overbrace{1}^{\to{}1}}{\underbrace{h^{2/3}}_{\to{}0}}$$

que claramente no existe (puesto que el denominador tiende a cero y el numerador no lo hace). Podemos concluir entonces que $f$ no es derivable en $x = 0$. La función es continua, su gráfica no es angulosa, pero la pendiente de las secantes se hace arbitrariamente grande cuando nos acercamos a $0$.


***

La comprobación de la derivabilidad o no de una función en determinado conjunto es importante para los temas que siguen. Esencialmente lo que haremos es estudiar la derivada para conocer la función. El instrumento para ello es el Teorema del valor medio que requiere entre sus hipótesis la derivabilidad de la función. Esto es, podemos usar el teorema y sus consecuencias solamente si la función es derivable.

**Derivada de una funci ́on definida a trozos**

Supongamos que queremos estudiar la derivabilidad de una función definida a trozos en un punto de pegado. Si tenemos:

$$f(x)=\begin{cases}g(x) & \text{si }x\leq{}a \\ k(x) & \text{si }x>a \end{cases}$$

donde $g(x)$ es derivable para $x < a$ y $k(x)$ es derivable para $x > a$. Entonces podemos afirmar que:

$$f'(x)=\begin{cases}g'(x) & \text{si }x<a \\ k'(x) & \text{si }x>a \end{cases}$$

¿Qué pasa en $x = a$? Esta pregunta la podemos responder de dos maneras.

**1° Forma**: Consiste en plantear el cociente incremental de $f(x)$ en $x=a$ e investigar si el limite de ese cociente existe. Eso implicará el estudio de los limites laterales:

$$\lim_{x\to{}a^+}\frac{f(x)−f(a)}{x−a}=\lim_{\substack{x\to{}a \\ x>a}}\frac{k(x)−g(a)}{x −a}$$

$$\lim_{x\to{}a^-}\frac{f(x)−f(a)}{x−a}=\lim_{\substack{x\to{}a \\ x<a}}\frac{g(x)−g(a)}{x −a}$$

Si esos limites existen y son iguales, la función será derivable en $a$ y el valor del limite será el de la derivada $f'(a)$.

**2° Forma**: Otra manera de hacerlo es verificar que la función es continua en $x=a$ y que las funciones $k$ y $g$ llegan a $a$ con la misma pendiente. Es decir, este procedimiento implica verificar dos condiciones:

1. $f$ es continua en $x = a$.

2. $\displaystyle{\lim_{x\to{}a^+}}k'(x)=\displaystyle{\lim_{x\to{}a^−}}g'(x)$

Si estas dos condiciones se cumplen, $f$ es derivable en $a$ y el valor de la derivada $f'(a)$ será el valor del limite obtenido en el punto 2.

### **4.2 El teorema del valor medio**

Este teorema, formulado originalmente por el matem ́atico Lagrange (1736-1813),
permite obtener informaci ́on sobre una funci ́on estudiando su derivada. El mismo
establece una relaci ́on entre la variaci ́on media de una funci ́on en un intervalo y la
variaci ́on instant ́anea en un punto interior. Concretamente, enunciamos:

***
$\color{blue}\text{Teorema}$:

**Valor medio o de Lagrange**: Sea $f(x)$ una función continua en un intervalo cerrado $[a, b]$ y que es derivable en el intervalo abierto $(a, b)$. Entonces existe un valor $c$ en el intervalo $(a, b)$ tal que se verifica:

$$\frac{f(b)−f(a)}{b −a}=f'(c)$$
***

Gráficamente tenemos:



El teorema nos dice que, para una función derivable, hay una tangente por un punto interior cuya pendiente es la misma que la de la recta secante por los extremos.

***
$\color{blue}\text{Ejemplo}$:

Como ejemplo del uso que podemos dar al teorema anterior demostremos el siguiente hecho, que es por cierto bastante intuitivo:

* Supongamos que $f(x)$ es una función continua en un intervalo cerrado $I =[a, b]$ y derivable en el interior del mismo. Entonces, si $f'(x) = 0$ para todo $x\in{}(a, b)$ se tiene que $f(x)$ es constante en $I$.\
Para verlo, consideremos un $x > a$ en $I$. El teorema anterior nos asegura que existe un $c$ en el interior de $I$ tal que:$$\frac{f(x)−f(a)}{x−a}=f'(c)=0\text{ por nuestra hipótesis}$$por lo tanto debe ser

$$f(x) −f(a) = 0$$
$$f(x) = f(a)$$

* Una consecuencia de la proposición anterior es que si dos funciones, digamos $f(x)$ y $g(x)$ tienen la misma derivada en un intervalo entonces difieren en una constante (ver al margen una interpretación gráfica de este hecho). En efecto, supongamos que ambas funciones son continuas en un intervalo $I$ como el de antes y derivables en el interior de $I$. Consideremos a la función diferencia $f(x) −g(x)$; su derivada es la diferencia de las derivadas, y como éstas son iguales, será nula en el interior de $I$. Por el punto anterior se tendrá: 

$$f(x) −g(x) = k \text{ constante}$$


***

### **4.3 Crecimiento y decrecimiento**

Veamos cómo el teorema del valor medio nos permite conocer el comportamiento de una funci ́on estudiando su derivada. En primer lugar, demos un par de definiciones

***
$\color{cyan}\text{Definición}$:

Sea $f(x)$ una función y sea $I$ un intervalo contenido en el dominio de $f(x)$.

Diremos que $f(x)$ es **creciente** en $I$ si cada vez que $x_1 < x_2$ se tiene que $f(x_1) < f(x_2)$, cualesquiera sean $x_1$ y $x_2$ en el intervalo $I$.

Diremos que $f(x)$ es **decreciente** en $I$ si cada vez que $x_1 < x_2$ se tiene que $f(x_1) > f(x_2)$, cualesquiera sean $x_1$ y $x_2$ en el intervalo $I$.
***

Hay cierta relación entre el hecho de que la función sea creciente y la derivada sea positiva, asi como entre función decreciente y derivada negativa. Precisaremos esa relación en el teorema siguiente.

***
$\color{blue}\text{Teorema}$:

Sea $f(x)$ una función continua en un intervalo $[a, b]$ y derivable en $(a, b)$.\
Entonces:

1. Si $f'(x) > 0$ en $(a, b)$ entonces $f(x)$ es creciente en $[a, b]$.

2. Si $f'(x) < 0$ en $(a, b)$ entonces $f(x)$ es decreciente en $[a, b]$
***

*Demostración*. Supongamos que $f'(x) > 0$ en $(a, b)$ , y sean $x_1 < x_2$ dos numeros en $[a, b]$. Queremos demostrar que $f(x_1) < f(x_2)$. Como $[x_1, x_2]\subset{}[a, b]$ resulta que $f(x)$ es continua en $[x_1, x_2]$ y derivable en $(x_1, x_2)$. El teorema del valor medio nos asegura la existencia de un numero $c$ perteneciente a $(x_1, x_2)$ tal que:

$$\frac{f(x_2) −f(x_1)}{x_2−x_1}=f'(c)$$

Ahora bien, como $f'(x) > 0$ en $(a, b)$ resulta que $f'(c) > 0$ y por lo tanto

$$\frac{f(x_2)−f(x_1)}{x_2−x_1}>0$$

El denominador de este último cociente es positivo puesto que $x_1<x_2$, y por lo tanto el numerador también es positivo, o sea:

$$f(x2) −f(x1) > 0 \text{ y por lo tanto } f(x2) > f(x1)$$

como queriamos demostrar. Un razonamiento similar se hace para el caso $f'(x) < 0$.

El teorema anterior permite reducir el estudio de los intervalos de crecimiento y decrecimiento de una función derivable en un intervalo cualquiera (finito o infinito, abierto, semiabierto o cerrrado) al estudio del signo de la derivada de la función en el interior del intervalo.

***
$\color{blue}\text{Ejemplo}$:

Determinar los intervalos de crecimiento y decrecimiento de la función $f(x)=x^3−4x^2$.

**Paso 1**. Estudiamos la derivabilidad y calculamos la derivada. Nuestra función es un polinomio, y por lo tanto es derivable en toda la recta. Su derivada es un polinomio de grado 2 y su gráfica, por lo tanto, es una parábola:

$$f'(x)=3x^2−8x$$

que es continua en toda la recta.

**Paso 2**. Encontramos los ceros de la derivada. Debemos resolver la ecuación

$$3x^2−8x=0$$

$$x(3x−8)=0$$

las soluciones son $x = 0$ y $x=\frac{8}{3}$.

**Paso 3**. Identificamos los intervalos donde conserva el signo y evaluamos en valores de prueba. La derivada es continua en todos los reales y se anula únicamente en $0$ y $8/3$. Por lo tanto $f'$ conserva el signo en los intervalos $(−\infty{}, 0)$, $(0, 8/3)$, $(8/3, +\infty{})$.

Una tabla ordena la información

| Intervalo | $(-\infty{},0)$ | $(0,8/3)$ | $(8/3,+\infty{})$ |
|:-:|:-:|:-:|:-:|
| $v_p$ | -1 | 2 | 3 |
| $f'(v_p)$ | $f'(-1)=11$ | $f'(2)=-4$ | $f'(3)=3$ |
| Signo de $f'$ | $+$ | $-$ | $+$ |
| Crecimiento de $f$ | Creciente $\nearrow$ | Decreciente  $\searrow$ | Creciente $\nearrow$ |

***

En vista de lo anterior, al estudiar intervalos de crecimiento y decrecimiento debemos ser cuidadosos en asegurarnos que estamos trabajando dentro de un intervalo en el que se verifican las hipótesis del teorema: continuidad hasta el borde (en el caso de un intervalo cerrado) y derivabilidad en el interior. El procedimiento debe ser:

* Excluir los puntos de discontinuidad de la función.

* Excluir los ceros y los puntos de discontinuidad de la derivada (algunos pueden coincidir con los anteriores).

* Estudiar el signo de la derivada en los intervalos determinados por las exclusiones anteriores.

### **4.4 Extremos locales**

Desde el comienzo del curso hemos dicho que uno de nuestros intereses mayores es determinar exactamente dónde una función alcanza un valor máximo o un valor mínimo. Ahora estamos en condiciones de avanzar en ese sentido. Empezamos dando dos definiciones:

***
$\color{cyan}\text{Defenición}$:

Sea $c$ un número en el dominio de una función $f(x)$.

Diremos que $f(x)$ alcanza en $c$ un **máximo local** si $f(c)\geq{}f(x)$ para todo $x$ en un intervalo alrededor de $c$.

Diremos que $f(x)$ alcanza en $c$ un **mínimo local** si $f(c)\leq{}(x)$ para todo $x$ en un intervalo alrededor de $c$.

En cualquiera de los casos anteriores diremos que $f(x)$ alcanza en $c$ un **extremo local**.
***

Tenemos el siguiente criterio para extremos locales. El criterio establece que, cuando $f(x)$ es una función derivable, por los puntos de la gr áfica en los que hay un extremo local pasa una tangente horizontal.

***
$\color{blue}\text{Teorema}$:

Supongamos que $f(x)$ es una función derivable en un valor $c$ y que $f(x)$ alcanza en $c$ un extremo local. Entonces $f'(c) = 0$.
***

*Demostración*. Supongamos que en $c$ se alcanza un m áximo local. Entonces, para $h$ suficientemente pequeño se tendrá:

$$f(c+h)\leq{}f(c)$$
$$f(c+h)−f(c)\leq{}0$$

Para el cociente incremental tendremos:

$$\frac{f(c+h)−f(c)}{h}\leq{}0\text{ si }h\text{ es positivo}$$
$$\frac{f(c + h) −f(c)}{h}\leq{}0\text{ si }h\text{ es negativo}$$

De donde podemos afirmar que:

$$\lim_{h\to{}0^+}\frac{f(c+h)−f(c)}{h}\leq{}0$$
$$\lim_{h\to{}0^−}\frac{f(c+h)−f(c)}{h}\leq{}0$$

Como $f(x)$ es derivable en $c$, ambos limites existen y son iguales a $f'(c)$. Por lo tanto debe ser $f'(c) = 0$ como queriamos demostrar.

#### **4.4.1 Números criticos**

***
$\color{cyan}\text{Definición}$:

Un numero $c$ se dice un numero crítico de una función $f(x)$, si es un punto de continuidad y sucede alguna de las siguientes cosas:

$f(x)$ es derivable en $c$ y $f'(c)=0$, o bien $f(x)$ no es derivable en $c$.
***



En ambos casos, $c$ es un numero critico de $f(x)$

De acuerdo al teorema anterior podemos enunciar la siguiente propiedad:

> **Si una función continua alcanza un extremo local en un número $c$ entonces $c$ es un número critico.**

**¿Cuándo en un número crítico hay un extremo local?** Y en tal caso, ¿cuándo hay un máximo local y cuándo hay un mínimo local? Las respuestas a estas preguntas van de la mano del estudio de los intervalos de crecimiento y decrecimiento por medio del signo de la derivada. En efecto si $c$ es un punto de anulación de la derivada pueden pasar dos cosas:

* La derivada cambia de signo al pasar por $c$. En este caso hay un extremo local, puesto que cambia el sentido de crecimiento de la función.\
Ejemplo: $f(x) = x^2$ en $x = 0$.

* La derivada no cambia de signo al pasar por c. En este caso no hay extremo local. La funci ́on no cambia el sentido de crecimiento aunque tiene una tangente horizontal.\
Ejemplo: $f(x) = x^3$ en $x = 0$.

***
$\color{blue}\text{Ejemplo}$:

En el ejemplo de la ***4.3*** estudiamos los intervalos de crecimiento y decrecimiento de $f(x)=x^3−4x^2$. Vimos que la función es derivable en toda la recta y determinamos los ceros de la derivada, obteniendo por tanto los números críticos $x=0$ y $x=8/3$. En ambos casos comprobamos que la derivada cambia de signo:

| Intervalo | $(-\infty{},0)$ | $(0,8/3)$ | $(8/3,+\infty{})$ |
|:-:|:-:|:-:|:-:|
| $v_p$ | -1 | 2 | 3 |
| $f'(v_p)$ | $f'(-1)=11$ | $f'(2)=-4$ | $f'(3)=3$ |
| Signo de $f'$ | $+$ | $-$ | $+$ |
| Crecimiento de $f$ | Creciente $\nearrow$ | Decreciente  $\searrow$ | Creciente $\nearrow$ |

Esta tablita nos permite concluir que en $x=0$ se alcanza un máximo local ($f$ es creciente antes y decreciente después $\nearrow{}\searrow{}$), y que en $x = 8/3$ se alcanza un mínimo local ($f$ es decreciente antes y creciente después $\searrow{}\nearrow{}$). En estos casos decimos que el valor del máximo local es $f(0) = 0$ y el valor del mínimo local es $f(8/3) = −256/27$.

### **4.5 Concavidad**

Hemos visto que el signo de la derivada determina el crecimiento de la función.\
Veremos ahora otro aspecto que es importante a la hora de conocer la forma de la gráfica de una función y cómo estudiarlo. Observemos las gráficas siguientes

Ambas funciones son crecientes en el intervalo $[a, b]$, sin embargo sus gráficas aparecen diferentes en algún sentido.

En la gráfica de la izquierda, desde una perspectiva geométrica, cada vez que trazamos la tangente por un punto la gráfica queda - por lo menos cerca de ese punto - *por debajo* de la tangente. De la misma manera, en la gráfica de la derecha la gráfica queda *por arriba* de la tangente.

En el primer caso diremos que la gráfica es **cóncava hacia abajo** en $[a, b]$, y en el segundo diremos que es **cóncava hacia arriba** en $[a, b]$.

Observemos que en el primer caso las pendientes de las tangentes son decrecientes, mientras que en el segundo caso las pendientes son crecientes. O sea, la concavidad está relacionada con el sentido de crecimiento de la derivada y, por lo tanto, con el signo de la derivada segunda (como lo sugiere el caso de la posición, la velocidad La derivada $f'(x)$ de una función $f(x)$ dada, es una nueva y la aceleraci ́on). Concretamente usaremos - sin demostración - el siguiente criterio para la concavidad.

> La derivada $f'(x)$ de una función $f(x)$ dada, es una nueva función. Si $f'(x)$ también es derivable, a su derivada la llamamos **derivada segunda** y la notamos $f''(x)$.

***
$\color{blue}\text{Teorema}$:

Si para todo $x$ en un intervalo $I$ se tiene que $f''(x)>0$, entonces la gráfica de $f(x)$ es **cóncava hacia arriba** en $I$.

Si para todo $x$ en un intervalo $I$ se tiene que $f''(x)<0$, entonces la gráfica de $f(x)$ es **cóncava hacia abajo** en $I$.
***

***
$\color{cyan}\text{Definición}$:

Un punto de continuidad de la gráfica de la función en el cual la concavidad cambia de sentido se denomina un **punto de inflexión**.


***

***
$\color{blue}\text{Ejemplo}$:

1. Consideremos una función cuadrática general $f(x)=ax^2+bx+c$. Las derivadas son $$f'(x)=2ax+b$$ $$f''(x)=2a$$
Por lo tanto, tendremos que la gráfica es cóncava hacia arriba cuando $a>0$ y cóncava hacia abajo cuando $a<0$, como ya se había mencionado en la Sección ***1.5***.

2. Estudiemos la concavidad de la función del Ejemplo de la Sección ***4.3***: $f(x)=x^3−4x^2 $. Sus derivadas son: $$f'(x)=3x^2−8x$$ $$f''(x)=6x−8$$
La segunda derivada es continua en toda la recta - de hecho es lineal - y se anula solamente en $x = 4/3$. Por lo tanto $f''$ conserva el signo en los intervalos $(−\infty{}, 4/3)$, $(4/3, +\infty{})$. Por este motivo, alcanza con tomar un valor de prueba en cada intervalo. Una tabla ordena la información.

| Intervalo | $(-\infty{},4/3)$ | $(4/3,+\infty{})$ |
|:-:|:-:|:-:|
| $v_p$ | 0 | 3 |
| $f''(v_p)$ | $f''(0)=-8$ | $f''(3)=10$ |
| Signo de $f'$ | $-$ | $+$ |
| Concavidad de $f$ | cóncava hacia abajo $\cap$ | cóncava hacia arriba $\cup$ |

Como la función es continua en $x=4/3$ y su gráfica cambia de concavidad allí, el punto $(4/3, f(4/3))=(4/3,−128/27)$ es un punto de inflexión.

Observen que los intervalos en los cuales se estudi ́o el signo
de $f''$ no son los mismos que los usados para estudiar el signo de $f'$.

### **4.6 Comportamiento en el infinito**

Lo visto hasta ahora (continuidad, asintotas verticales, extremos locales, intervalos de crecimiento, concavidad) nos permite hacernos una idea bastante aproximada de la forma que tendrá la gráfica de una función. Un último aspecto que consideraremos es el "comportamiento en el infinito" de la función. Con esto nos referimos al comportamiento de los valores de la función cuando la variable se hace grande y positiva o grande y negativa. Concretamente:

***
$\color{cyan}\text{Defición}$:

* Diremos que $\displaystyle{\lim_{x\to{}+\infty{}}}f(x)=L$ si los valores de $f(x)$ están tan próximos a $L$ como queramos, tomando $x$ suficientemente grande y positivo.

* Diremos que $\displaystyle{\lim_{x\to{}-\infty{}}}f(x)=L$ si los valores de $f(x)$ están tan próximos a $L$ como queramos, tomando $x$ suficientemente grande y negativo

En cualquiera de los casos anteriores diremos que la recta de ecuación $y = L$ es una **asíntota horizontal** de la gráfica de $f(x)$.
***

***
$\color{red}\text{Ejemplo}$:

Determinar la existencia de as ́ıntotas horizontales para cada una de las siguientes funciones.

1. $f(x)=\frac{1}{x}$\
Calculamos los l ́ımites de la funci ́on cuando $x\to{}\pm{}\infty{}$ $$\lim_{x\to{}+\infty{}}f(x)=\lim_{x\to{}+\infty{}}\frac{\overbrace{1}^{\to{}1}}{\underbrace{x}_{\to{}+\infty{}}}=0\text{ y }\lim_{x\to{}-\infty{}}f(x)=\lim_{x\to{}-\infty{}}\frac{\overbrace{1}^{\to{}1}}{\underbrace{x}_{\to{}-\infty{}}}=0$$
Podemos concluir que la  ́unica asintota horizontal de la gráfica es la recta $y = 0$.

2. $f(x)=\frac{2x−1}{x}$\
Para estudiar el comportamiento en el infinito conviene escribir $$f(x)=\frac{2x−1}{x}=\frac{2x}{x}−\frac{1}{x}=2−\frac{1}{x}$$
Entonces $$\lim_{x\to{}\pm{}\infty{}}f(x)=\lim_{x\to{}\pm{}\infty{}}2−\frac{1}{x}=2$$
La recta $y = 2$ resulta la única asintota horizontal de la gráfica de $f(x)$.

3. $f(x) = x^2$\
En este caso los valores de la función se hacen grandes y positivos, tanto para $x$ grande y positivo como para $x$ grande y negativo.\
Cuando esto sucede lo indicamos así $$\lim_{x\to{}+\infty{}}f(x)=+\infty{}$$ $$\lim_{x\to{}-\infty{}}f(x)=+\infty{}$$
Podemos concluir que la gráfica de $f(x)$ no posee asíntotas horizontales.
***

Para estas últimas expresiones valen las advertencias hechas en el caso del comportamiento asintótico: son maneras de describir el comportamiento de la función cuando la variable se hace grande y positiva o grande y negativa y no se puede operar con estos símbolos como si fueran números.

Analizaremos el comportamiento en el infinito de algunas de las funciones más comunes.

#### **4.6.1 Comportamiento en el infinito de un polinomio**

1. Supongamos que tenemos un monomio, esto es un polinomio de un único término $$p(x)=ax^n$$
Un breve análisis muestra que el comportamiento en el infinito de este monomio depende del signo del coeficiente $a$ y de que el exponente $n$ sea par o impar. Resumimos en una tabla

| # | # | $\displaystyle{\lim_{x\to{}+\infty{}}}ax^n$ | $\displaystyle{\lim_{x\to{}-\infty{}}}ax^n$ |
|:-:|:-:|:-:|:-:|
| $n$ par | $a > 0$ | $+\infty{}$ | $+\infty{}$ |
| $n$ par | $a < 0$ | $-\infty{}$ | $-\infty{}$ |
| $n$ impar | $a > 0$ | $+\infty{}$ | $-\infty{}$ |
| $n$ impar | $a < 0$ | $-\infty{}$ | $+\infty{}$ |


2. Consideremos ahora, por ejemplo, el polinomio $p(x)=−5x^5+2x^4−x^3+1$.\
El limite $$\lim_{x\to{}\pm{}\infty{}}p(x)=\lim_{x\to{}\pm{}\infty{}}−5x^5+2x^4−x^3+1$$
no es posible calcularlo directamente porque aparecen indeterminaciones. Si
sacamos factor común $−5x^5$ en $p(x)$ tendremos $$p(x) = −5x^5 \left( 1+\frac{2x^4}{−5x^5}−\frac{x^3}{−5x^5}+\frac{1}{−5x^5} \right) = −5x^5 \left( 1+\frac{2}{−5}\cdot{}\frac{1}{x}−\frac{1}{−5}\cdot{}\frac{1}{x^2}+\frac{1}{−5}\cdot{}\frac{1}{x^5}\right)$$
En esta última expresión, el factor entre paréntesis tiende a $1$ cuando $x\to{}\pm{}\infty{}$, por lo tanto $$\lim_{x\to{}+\infty{}}p(x)=\lim_{x\to{}+\infty{}} \overbrace{−5x^5}^{\to{}-\infty{}} \overbrace{\left( 1+\overbrace{\frac{2}{−5}\cdot{}\frac{1}{x}}^{\to{}0}−\overbrace{\frac{1}{−5}\cdot{}\frac{1}{x^2}}^{\to{}0}+\overbrace{\frac{1}{−5}\cdot{}\frac{1}{x^5}}^{\to{}0}\right)}^{\to{}1} = -\infty{}$$ $$\lim_{x\to{}-\infty{}}p(x)=\lim_{x\to{}-\infty{}} \overbrace{−5x^5}^{\to{}+\infty{}} \overbrace{\left( 1+\overbrace{\frac{2}{−5}\cdot{}\frac{1}{x}}^{\to{}0}−\overbrace{\frac{1}{−5}\cdot{}\frac{1}{x^2}}^{\to{}0}+\overbrace{\frac{1}{−5}\cdot{}\frac{1}{x^5}}^{\to{}0}\right)}^{\to{}1} = +\infty{}$$
De manera que el comportamiento en $\pm{}\infty{}$ del polinomio es el comportamiento de su monomio de mayor grado, es decir, de $−5x^5$. Este resultado es general, puesto que el mismo procedimiento puede usarse con cualquier polinomio $p(x) = a_nx^n+a_{n−1}x^{n−1}+\dotsc{}+ a_1x + a_0$.

#### **4.6.2 Comportamiento en el infinito de una función racional**

Una función racional es un cociente de polinomios. Pongamos $r(x)=\frac{p(x)}{q(x)}$, donde $p(x)$ y $q(x)$ son polinomios. Nos interesa el comportamiento de $r(x)$ cuando $x\to{}\pm{}\infty{}$. A menos que $p(x)$ o $q(x)$ sean constantes tenemos un cociente de dos funciones, ambas tendiendo a $\pm{}\infty{}$ y, por lo tanto, es uno de los casos en los que el limite está indefinido. Para definir la situación procedemos con $p(x)$ y $q(x)$ como lo hicimos en el ejemplo anterior. Lo ilustramos con un ejemplo:

Consideramos a la función racional $r(x)=\frac{−2x^2+x−1}{x^3+3}$. Si operamos con cada polinomio sacando factor común el monomio de mayor grado tenemos

$$r(x)=\frac{-2x^2 \left( 1 + \frac{1}{-2x} - \frac{1}{-2x^2} \right)}{x^3 \left( 1 + \frac{3}{x^3} \right)} = \frac{-2x^2}{x^3} \cdot{} \frac{\left( 1 + \frac{1}{-2x} - \frac{1}{-2x^2} \right)}{\left( 1 + \frac{3}{x^3} \right)} = \frac{-2}{x} \cdot{} \frac{\left( 1 + \frac{1}{-2x} - \frac{1}{-2x^2} \right)}{\left( 1 + \frac{3}{x^3} \right)}$$

En esta última expresión, ambos paréntesis tienden a $1$ cuando $x\to{}\pm{}\infty{}$. En consecuencia

$$\lim_{x\to{}+\infty{}} r(x) = \lim_{x\to{}+\infty{}} \overbrace{\frac{-2}{x}}^{\to{}0} \cdot{} \frac{\overbrace{\left( 1 + \frac{1}{-2x} - \frac{1}{-2x^2} \right)}^{\to{}1}}{\underbrace{\left( 1 + \frac{3}{x^3} \right)}_{\to{}1}} = 0$$

$$\lim_{x\to{}-\infty{}} r(x) = \lim_{x\to{}-\infty{}} \overbrace{\frac{-2}{x}}^{\to{}0} \cdot{} \frac{\overbrace{\left( 1 + \frac{1}{-2x} - \frac{1}{-2x^2} \right)}^{\to{}1}}{\underbrace{\left( 1 + \frac{3}{x^3} \right)}_{\to{}1}} = 0$$

De manera que el comportamiento en $\pm{}\infty{}$ de la funcin racional es el comportamiento del cociente de los dos monomios de mayor grado, es decir, de $\frac{−2x2}{x3}=\frac{−2}{x}$.

Este resultado es general, puesto que el mismo procedimiento puede usarse con cualquier función racional $r(x)=\frac{p(x)}{q(x)}$. Esto es, si

$$p(x) = a_nx^n + a_{n−1}x^{n−1} + \dotsc{} + a_1x + a_0$$
$$q(x) = b_mx^m + b_{m−1}x^{m−1} + \dotsc{} + b_1x + b_0$$

entonces el limite $\displaystyle{\lim_{x\to{}\pm{}\infty{}}r(x)}$ queda determinado por el comportamiento en $\pm{}\infty{}$ del cociente $\frac{a_nx^n}{b_mx^m}$. Hay tres posibilidades para este cociente:

1. $n > m$. El cociente es el monomio $\frac{a_n}{b_m}\cdot{}x^{n−m}$ que tiende a $\pm{}\infty{}$ depeniendo del signo del coeficiente y de la paridad de $n −m$.

2. $n < m$. El cociente es de la forma $\frac{a_n}{b_m}\cdot{}\frac{1}{x^{m−n}}$ que tiende a $0$. En este caso la recta $y = 0$ es una asintota horizontal de la gráfica de $r(x)$.

3. $n = m$. El cociente es de la forma $\frac{a_n}{b_m}$ (constante) y por tanto ése es el limite.\ En este caso la recta $y=\frac{a_n}{b_m}$ es una asintota horizontal de la gráfica de $r(x)$.

### **4.7 Estudio de una función racional**

El estudio de una función es una actividad de cierta complejidad, que integra los distintos conceptos que hemos visto en la Sección. Ese estudio concluye con la realización de una gráfica que contenga los elementos determinados previamente. El estudio debe incluir siempre los siguientes pasos:

1. **Estudio del dominio**. Descripción del dominio en terminos de intervalos.

2. **Estudio de la continuidad**. Esto incluye

    * Intervalos de continuidad.

    * Puntos de discontinuidad. Clasificación de discontinuidades en dichos puntos mediante el estudio de la existencia del limite.

    * Existencia de asintotas verticales en los puntos de discontinuidades inevitables; esto surge del item anterior.

3. **Comportamiento en el infinito**. Existencia de asintotas horizontales.

4. **Primera derivada**. Esto incluye

    * Cálculo de la derivada donde la misma existe e identificación de los puntos donde no existe.

    * Ceros de la derivada. Estudio del signo de la derivada y determinación de los intervalos de crecimiento y decrecimiento.

    * Determinación de extremos locales, para lo cual se estudian **todos** los números criticos.

5. **Segunda derivada**. Debe realizarse

    * Cálculo de la segunda derivada donde la misma existe e identificación de los puntos donde no existe.

    * Estudio del signo de la segunda derivada y determinación de los intervalos de concavidad. Puntos de inflexión.

6. **Puntos notables**. Intersección con los ejes $x$ e $y$.

7. **Construcción de una tabla de valores** que incluya a los puntos criticos y puntos de inflexión. 

8. **Dibujo de la gráfica** en el cual se señaalen las caracteristicas principales y que sea coherente con el estudio realizado.

#### **4.7.1 Forma estándar de una función racional**

Algunas funciones racionales pueden expresarse en una forma que facilita su estudio, simplificando las cuentas y proporcionando una idea previa de cómo puede ser su gráfica. Describamos el método y después lo ilustramos con algunos ejemplos.

Supongamos que se tiene la función racional

$$f(x)=\frac{p(x)}{q(x)}$$

y además que el grado del numerador es mayor o igual que el grado del denominador.\
Podemos entonces dividir el polinomio $p(x)$ por el polinomio $q(x)$ obteniendo un cociente $c(x)$ y un resto $r(x)$ que verifican lo siguiente

$$p(x)=c(x)q(x)+r(x)\text{ con }r(x)=0\text{ o grado }r(x)<\text{ grado de }q(x)$$

Esto nos permite escribir a esta función racional de la siguiente manera

$$f(x)=\frac{p(x)}{q(x)}=\frac{c(x)q(x)+r(x)}{q(x)}=\frac{c(x)q(x)}{q(x)}+\frac{r(x)}{q(x)}$$
$$f(x)=\overbrace{c(x)}^{\text{COCIENTE}}+\underbrace{\overbrace{\frac{r(x)}{q(x)}}^{\text{RESTO}}}_{\text{DENOMINADOR}}$$

A esta manera de expresar a $f(x)$ la llamamos **forma estándar**. El primer término es la **parte polinomial** y el segundo es la **parte fraccionaria**:

$$f(x)=\underbrace{\overbrace{c(x)}^{\substack{\text{PARTE} \\ \text{POLINOMINAL}}}+\overbrace{\frac{r(x)}{q(x)}}^{\substack{\text{PARTE} \\ \text{FRACCIONARIA}}}}_{\text{FORMA ESTÁNDAR}}$$

***
$\color{red}\text{Ejemplo}$:

Escribamos en la forma estándar a la función racional $f(x)=\frac{x^3−x^2−1}{x −1}$.

Vemos que el grado del numerador es mayor que el del denominador, y entonces hacemos la división

$$
\begin{array}{rll | l}
x^3 & -x^2 & -1 & x - 1 \\ \hline
x^3 & -x^2 &    & x^2 \quad \text{(COCIENTE)} \\ \hline
0   & 0    & -1 & \\ 
    &      & \text{(RESTO)} & 
\end{array}
$$

Podemos en consecuencia escribir

$$f(x)=\underbrace{\overbrace{x^2}^{\substack{\text{PARTE} \\ \text{POLINOMINAL}}}+\overbrace{\frac{-1}{x-1}}^{\substack{\text{PARTE} \\ \text{FRACCIONARIA}}}}_{\text{FORMA ESTÁNDAR}}$$
***

La forma estándar tiene dos ventajas. La primera es que en muchos casos simplifica los cálculos, al bajar el grado del numerador en la parte fraccionaria. Esto no es menor: en el ejemplo anterior la derivada usando la forma estándar puede calcularse así:

$$f'(x)=2x+(x −1)^{−2}$$

mientras que la versión original hay que derivarla usando la regla del cociente, la cual origina la expresión:

$$f'(x)=\frac{(3x^2-2x)(x-1)-(x^3-x^2-1)}{(x-1)^2}$$

que si bien es igual a la anterior, es mucho más complicada para manipular.

La segunda ventaja es que expone el comportamiento asintotico de la función. En efecto tenemos:

* En el infinito el comportamiento de la función racional es el de la parte polinomial, puesto que la parte fraccionaria tiende a $0$. Esto lo indicamos así

$$f(x)\approx{}c(x)\text{ para }x\text{ grande }(\text{es decir, cuando }x\to{}\pm{}\infty{})$$

* En las discontinuidades inevitables el comportamiento asintotico es el de la parte fraccionaria (puesto que la parte polinomial es continua siempre). Si la discontinuidad inevitable está en $x=x_0$, este comportamiento lo expresamos asi

$$f(x)\approx{}\frac{r(x)}{q(x)}\text{ para }x\text{ próximo a }x_0 (\text{es decir, cuando }x\to{}x_0^{\pm{}})$$

Con referencia al ejemplo anterior, estos hechos los expresamos de la siguiente manera:

* Para $x$ grande, $f(x)\approx{}x^2$ (leemos "efe se comporta como equis cuadrado").

* Para $x$ próximo a $1$, $f(x)\approx{}\frac{−1}{x −1}$ (leemos: "efe se comporta como menos uno sobre equis menos uno")

Si graficamos la parte polinomial $(y=x^2)$ y la parte fraccionaria $(y=\frac{−1}{x−1})$ esas gráficas nos dirán cómo es la gráfica de $f(x)$ para $x$ grandes (positivos y negativos) y para $x$ próximos a $1$.

Estudiemos la función racional $f(x)=\frac{x^4+1}{x^2}$.

Para estudiar la función, seguimos los pasos 1 a 8 descriptos en ***4.7.1***.

cuyas unicas raices reales son $x=1$ y $x=−1$\

### **4.8 Funciónes inversas**

#### **4.8.1 La imagen de una función**

***
$\color{cyan}\text{Definión}$:

Consideremos una función numerica $f(x)$ con un dominio $D$. Al conjunto de valores que la función $f(x)$ efectivamente alcanza se lo llama **imagen de $f$** y se lo denota $Im(f)$ o bien $f(D)$ cuando se quiere tener presente el dominio de la función. En la notación de conjuntos
$$Im(f)=f(D)=\{\text{ y/ existe un }x\in{}D\text{ tal que }f(x)=y\}$$
***

Gráficamente, la imagen puede verse como la proyección sobre el eje $y$ de la gráfica de la función $f(x)$.

No siempre es posible contar con la gráfica de una función para poder deducir cual es su imagen. En general, para determinar la imagen se usa la definición anterior.\
Veamos cómo hacerlo con el siguiente ejemplo.

***
$\color{red}\text{Ejemplo}$:

La cuestión de encontrar "a mano" la imagen de una función dada no es fácil. Implica determinar para qué valores de y la ecuación $f(x)=y$ tiene o no tiene solución. Por ejemplo, si consideramos a la función $f(x)=x^2+20$, ¿cuándo un valor y pertenece a la imagen de $f$? De acuerdo a la definición, cuando exista un $x$ tal que $f(x)=y$.\
Es decir, debe existir un $x$ tal que:
$$x^2+20=y$$
$$x^2=y−20$$
para que esta última ecuación admita solución real, debe ser $y−20\geq{}0$, o sea: $y\geq{}20$.\
En tal caso tendremos como soluciones $x=0$ para $y=20$; o bien $x=\pm{}\sqrt{y−20}$ para $y>20$. Por lo tanto, $Im(f)=[20, +∞)$. 
***

#### **4.8.2 La inversa de una función continua y monótona**

Supongamos ahora que $f(x)$ es una función continua y creciente definida en un intervalo $I=[a,b]$. Al ser $f$ continua el teorema del valor intermedio nos asegura que $Im(f)=[f(a),f(b)]=K$.



De manera que, para cada y en el intervalo $K$ existe un $x$ en el intervalo $I$ tal que $f(x)=y$. Además, el hecho de que $f$ sea creciente nos dice que ese $x$ es el  ́unico que por $f$ se aplica en $y$.

Ahora bien, si a cada y en $K$ le asociamos el único $x$ que se aplica en el, obtendremos una nueva función cuyo dominio es el intervalo $K$ y cuya imagen es el intervalo $I$. A esa función la llamaremos **función inversa** de $f$, y la denotaremos $f^{−1}$.

> *Atención: $\displaystyle{f^{−1}}$ es una notación para la función inversa, no confundir con $\displaystyle{\frac{1}{f}}$*.

Por la manera que hemos definido a $\displaystyle{f^{-1}}$ tenemos las siguientes propiedades:

* Para cualquier $x\in{}[a,b]$ se tiene que $f^{−1}(f(x))=x$

* Para cualquier $y\in{}[f(a),f(b)]$ se tiene que $f(f^{−1}(y))=y$

Estas dos propiedades sobre $f$ y $f^{−1}$ indican que una función deshace lo que hizo la otra.

Todo lo que hemos dicho para una función continua y creciente vale para una función $f$ continua y decreciente teniendo en cuenta que en ese caso la imagen de $f$ será el intervalo $[f(b),f(a)]$.

Para ahorrar palabras diremos que una función es **monótona** en un intervalo $I$ si es creciente en $I$ o es decreciente en $I$.

El haber considerado en la discusión anterior una función con dominio en un intervalo cerrado tuvo por finalidad poder identificar claramente a la imagen de la función. Si nuestra función es continua y monótona en un intervalo $I$ cualquiera (abierto o semiabierto; finito o infinito) la imagen $f(I)$ es también un intervalo pero no necesariamente del mismo tipo que $I$.

La aclaración anterior nos permite enunciar el siguiente resultado para un intervalo $I$ general:

***
$\color{blue}\text{Teorema}$:

Sea $f : I \to{} f(I)$ una función continua y monótona. Existe una función que llamamos **función inversa** de $f$ y que denotamos $f^{−1}$ cuyo dominio es $f(I)$ y cuya imagen es $I$ que verifica: 

* $f^{−1}$ es continua en $f(I)$,

* para cualquier $x\in{}I$ se tiene que $f^{−1}(f(x))=x$ y, además,

* para cualquier $y\in{}f(I)$ se tiene que $f(f^{−1}(y))=y$.
***

***
$\color{red}\text{Ejemplo}$:

Consideremos la función del ejemplo anterior $f(x)=x^2+20$. Estudiando el signo de la derivada $f'(x)=2x$ comprobamos que la función es decreciente en $I=(−\infty{}, 0]$ y creciente en $J=[0,+\infty{})$. Por lo tanto, no es posible en principio usar lo anterior para asegurar la existencia de una función inversa en todo el dominio de $f$. Pero si consideramos la **restricción** de $f$ al intervalo $I$, lo que no es otra cosa que mirar a
la misma función $f(x)=x^2+20$ pero con dominio en el intervalo $I$; esa restricción, que denotaremos $f_I$ si admitirá una función inversa $f^{−1}_{I}$.

De la misma manera, la restricción de $f$ al intervalo $J$ que denotaremos $f_J$ que es creciente, también admitirá una función inversa $f^{−1}_{J}$. Determinemos los dominios e im ́agenes de las funciones inversas.

1. El dominio de $f^{−1}_{I}$ es la imagen por $f$ del intervalo $I$ o sea $[20,+\infty{})$. De modo que $f^{−1}_{I}$: $f(I)=[20,+\infty{})\to{}I=(−\infty{}, 0]$. No siempre es posible determinar la expresión de la función inversa, pero en este caso si $f^{−1}_{I}(y)=x$ entonces $y=f(x)$ con $x$ perteneciente al intervalo $I=(−\infty{},0]$. Esto es: $$y=x^2+20$$ $$x^2=y −20\text{ con }x\text{ en }I=(−\infty{},0]$$ $$x = −\sqrt{y−20}$$
En consecuencia, $f^{−1}_{I}(y)=−\sqrt{y−20}$, o si queremos usar como variable independiente a $x$ será $f^{−1}_{I}(x)=−\sqrt{x−20}$

2. De la misma forma podemos trabajar con $f^{−1}_{J}$. Su dominio es $[20,+\infty{})$ mientras que su imagen es el intervalo $J=[0,+\infty{})$. Un cálculo similar al anterior nos da la expresión de $f^{−1}_{J}$:

$$f_{J}^{−1}(x)=\sqrt{x−20}$$
***

#### **4.8.3 Propiedades de la función inversa**

En lo que sigue sea $f:I\to{}J$ una función continua y monótona, donde $J=f(I)$.\
Por lo tanto existe una función inversa $f^{−1}:J\to{}I$.

**La gráfica de la inversa.**

Para que un punto $(x, y)$ esté en la gráfica de $f^{−1}$ debe ser $y=f^{−1}(x)$. Pero entonces $f(y)=x$ lo cual nos dice que el punto $(y, x)$ está en la gráfica de $f$.\
Usen la Figura 1 para mostrar que un punto de coordenadas $(a, b)$ es el simétrico del punto de coordenadas $(b, a)$ respecto de la recta $y=x$.



De lo anterior podemos concluir que las gráficas de $f$ y $f^{−1}$ son simétricas respecto de esa recta, como se ve en el dibujo del margen.



**La derivada de la función inversa.**

Supongamos ahora que $f$ es derivable en $x = a$, y sea $b = f(a)$.

1. Ecuacion de la recta tangente a la grafica de $f$ por el punto $(a,b)$ $$y-b=f'(a)\cdot{}(x-a)$$

2. Si en la ecuación anterior intercambiamos las variables queda la siguiente ecuación: $$x−b=f'(a)\cdot{}(y−a)$$ que es una recta, simétrica de la anterior respecto a la recta $y = x$ ¿Cuál es la pendiente de esa recta? $$y-a=\frac{1}{f'(a)}\cdot{}(x-b)$$
La pendiente de la recta de la funcion inversa es $\displaystyle{\frac{1}{f'(a)}}$

3. Gráficamente que si la recta $l_1$ es tangente a la gráfica de $f$ por el punto $(a, b)$ entonces la recta simétrica de $l_1$ respecto a $y = x$ es tangente a la gráfica de $f^{−1}$ por el punto $(b, a)$.

4. De los puntos anteriores podemos concluir que cuando $f'(a)\neq{}0$ la recta simétrica a la tangente no es vertical, y su pendiente es igual a $\displaystyle{\frac{1}{f'(a)}}$.\
Lo cual nos dice que si $b=f(a)$ entonces la derivada de $f^{−1}$ en $b$ es igual a $\displaystyle{\frac{1}{f'(a)}}$.

5. Comprobemos lo anterior "algebraicamente". Supongamos que $y=f(x)$. Tenemos la identidad $$f^{−1}(y)=x$$
si la derivamos usando la regla de la cadena se obtiene: $$[f^{−1}]'(y)\cdot{}y'=1$$ $$[f^{−1}]'(y) =\frac{1}{y'}$$
y si ponemos todo en términos de $x$: $$[f^{−1}]'(f(x))=\frac{1}{f'(x)}\text{ siempre que }f'(x)\neq{}0$$

**Resumen**

Resumimos lo visto hasta ahora sobre funciones inversas. Supongamos que $f:I\to{}f(I)$ es una función continua y monótona. Entonces:

* Existe una función inversa $f^{-1}:f(I)\to{}I$ que es continua

* Para cualquier $x\in{}I$ se tiene que $f^{−1}(f(x))=x$

* Para cualquier $y\in{}f(I)$ se tiene que $f(f^{−1}(y))=y$

* La gráfica de $f^{−1}$ es la simétrica de la gráfica de $f$ respecto de la recta $y=x$

* Si $f(a)=b$ y existe $f'(a)$ y es diferente de $0$ entonces $f^{−1}$ es derivable en $b$ y se tiene $[f−1]'(b)=\frac{1}{f'(a)}$
