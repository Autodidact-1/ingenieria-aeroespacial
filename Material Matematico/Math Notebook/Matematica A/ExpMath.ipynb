**Desarrollo de demostraciones**

Utiliza ecuaciones alineadas.

$$
\begin{aligned}
A
&=\sum_{i=1}^{n}\Delta x f(x_i^*)\\
\\
&=\sum_{i=1}^{n}\Delta x
f\!\left(a+\left(i-\frac12\right)\Delta x\right)\\
\\
&\xrightarrow[n\to\infty]{}
\int_a^b f(x)\,dx
\end{aligned}
$$

Es una forma muy elegante de presentar deducciones.

$$
\boxed{ \int_{a}^{b} f(x) \, dx = \lim_{n\to{}\infty{}} \sum_{i=1}^{n} f(x_i) \Delta{} x}
$$
